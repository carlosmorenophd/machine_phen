```python
import os
import sys
module_path = os.path.abspath(os.path.join('../..'))
if module_path not in sys.path:
    sys.path.append(module_path)
```

```python
from machines.predictions.linear_prediction import RidgePrediction, LinearRegressionPrediction, LassoPrediction
from preprocesses.preprocess import Preprocess
from preprocesses.enums import TransformEnum, TypeFileEnum, StandardScaleEnum
from metrics.error_metric import ErrorMetric
from metrics.enums import MetricEnum
```

Loading data

```python
file_csv = '/Users/yeiden/repo/machine_phen/data/phenotypic_obregon_151617.csv'
preprocessing =  Preprocess(file_name=file_csv, type_file=TypeFileEnum.CSV)
preprocessing.read_file(
    transform=TransformEnum.PASS,
    standard_scale=StandardScaleEnum.BASIC,
)
preprocessing.build_train_and_test()
x_train, y_train = preprocessing.get_train()
```

Run bayesian training

```python
linear =  LinearRegressionPrediction()
ridge = RidgePrediction()
lasso = LassoPrediction()
linear.training(x_train=x_train, y_train=y_train)
ridge.training(x_train=x_train, y_train=y_train, alpha=0.5)
lasso.training(x_train=x_train, y_train=y_train, alpha=0.9)
```

Predict and get metric

```python
x_test, y_test = preprocessing.get_test()
y_predicted_linear = linear.prediction(x_test=x_test)
y_predicted_ridge = ridge.prediction(x_test=x_test)
y_predicted_lasso = lasso.prediction(x_test=x_test)
```

```python
metric = ErrorMetric(y_predicted=y_predicted_linear, y_test=y_test)
metric.calculate_metric_prediction()
print('Root mean squared error -> {}'.format(metric.get_metric(metric=MetricEnum.ROOT_MEAN_SQUARED_ERROR)))
print('R2 -> {}'.format(metric.get_metric(metric=MetricEnum.R2_SCORE)))
print('MAPE -> {}'.format(metric.get_metric(metric=MetricEnum.MEAN_ABSOLUTE_PERCENTAGE_ERROR)))
metric.get_all_from_predict()
```

```python
metric = ErrorMetric(y_predicted=y_predicted_ridge, y_test=y_test)
metric.calculate_metric_prediction()
print('Root mean squared error -> {}'.format(metric.get_metric(metric=MetricEnum.ROOT_MEAN_SQUARED_ERROR)))
print('R2 -> {}'.format(metric.get_metric(metric=MetricEnum.R2_SCORE)))
print('MAPE -> {}'.format(metric.get_metric(metric=MetricEnum.MEAN_ABSOLUTE_PERCENTAGE_ERROR)))
metric.get_all_from_predict()
```

```python
metric = ErrorMetric(y_predicted=y_predicted_lasso, y_test=y_test)
metric.calculate_metric_prediction()
print('Root mean squared error -> {}'.format(metric.get_metric(metric=MetricEnum.ROOT_MEAN_SQUARED_ERROR)))
print('R2 -> {}'.format(metric.get_metric(metric=MetricEnum.R2_SCORE)))
print('MAPE -> {}'.format(metric.get_metric(metric=MetricEnum.MEAN_ABSOLUTE_PERCENTAGE_ERROR)))
metric.get_all_from_predict()
```
